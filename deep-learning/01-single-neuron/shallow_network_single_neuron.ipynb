```python
# A single neuron implemented from scratch using NumPy.

# Topics:
# - Forward pass
# - ReLU
# - Loss
# - Backpropagation
# - Gradient descent
```

```python
import numpy as np
import matplotlib.pyplot as plt


x = np.array([1, 2, 3, 4])
y = np.array([5, 3, 2, 1])


def ReLU(x):
    return np.maximum(0, x)


def MSE(y, y_hat):
    return np.mean((y_hat - y) ** 2)

w = 0.5
b = 0.0

learning_rate = 0.01
epochs = 1000

loss_history = []


for epoch in range(epochs):

    # Forward pass
    z = w * x + b
    y_hat = ReLU(z)

    # Loss
    loss = MSE(y, y_hat)
    loss_history.append(loss)

    # Gradients
    error = y_hat - y

    dw = np.mean(2 * error * x)
    db = np.mean(2 * error)

    # Gradient descent
    w = w - learning_rate * dw
    b = b - learning_rate * db
    if epoch ==1:
      print("loss initilly = ")
      print(loss)

    if epoch == 100000:
      print("loss initilly = ")
      print(loss)



# Final Results
print("Final w:", w)
print("Final b:", b)
print("Final predictions:", y_hat)
print("Final loss:", loss)

# Loss Curve
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.show()
```
